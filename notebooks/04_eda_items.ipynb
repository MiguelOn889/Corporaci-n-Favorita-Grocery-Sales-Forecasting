## Diccionario de Variables

| Columna | Descripción | Tipo de dato |
|----------|-------------|--------------|
| item_nbr | Identificador único del producto | `int64` |
| family | Familia o categoría a la que pertenece el producto | `str` |
| class | Identificador de la clase del producto | `int64` |
| perishable | Indica si el producto es perecedero (`1`) o no perecedero (`0`) | `int64` |

### Observaciones

- **item_nbr** es el identificador único de cada producto y será utilizado para relacionar este dataset con el historial de ventas (`train`).

- **family** es una variable categórica que agrupa los productos según su categoría comercial y permitirá analizar el comportamiento de las ventas por familias.

- **class** representa una clasificación más específica dentro de cada familia. Aunque está almacenada como `int64`, corresponde a una **variable categórica** y no debe interpretarse como un valor numérico continuo.

- **perishable** es una variable binaria codificada como `0` (producto no perecedero) y `1` (producto perecedero), la cual permitirá analizar diferencias en el comportamiento de venta entre ambos tipos de productos.

- El dataset no presenta valores nulos, por lo que inicialmente no se requieren procesos de limpieza relacionados con datos faltantes.

## Análisis de las variables numéricas

A continuación se presentan las principales observaciones obtenidas a partir de las estadísticas descriptivas de las variables numéricas del dataset **Items**.

### Observaciones

- **item_nbr**
  - Cada valor corresponde al identificador único de un producto del catálogo.
  - Aunque está almacenado como un número entero (`int64`), representa una **variable categórica** y no una magnitud numérica.
  - Su principal función es permitir la integración con el dataset de ventas (`train`).

- **class**
  - Corresponde a una clasificación más específica de los productos dentro de sus respectivas familias.
  - Aunque está codificada como `int64`, debe tratarse como una **variable categórica**, ya que sus valores representan códigos y no cantidades.
  - Esta variable podrá utilizarse para realizar análisis más detallados del comportamiento de grupos específicos de productos.

- **perishable**
  - Es una variable binaria cuyos valores son **0** (producto no perecedero) y **1** (producto perecedero).
  - Permite diferenciar los productos según su naturaleza, lo cual puede ser relevante para analizar patrones de venta, rotación e impacto de promociones.
  - Al tratarse de una variable binaria, no debe interpretarse mediante medidas como la media o la desviación estándar, sino como la proporción de productos perecederos y no perecederos.

### Conclusiones

- Las variables **`item_nbr`** y **`class`** deben tratarse como **variables categóricas** durante el análisis exploratorio y el modelado.

- La variable **`perishable`** es una variable binaria que aporta información relevante sobre el tipo de producto y podrá utilizarse como característica predictiva.

**No se observan inconsistencias en la estructura de las variables numéricas, por lo que el dataset presenta una base adecuada para continuar con el análisis exploratorio y su posterior integración con el historial de ventas**
