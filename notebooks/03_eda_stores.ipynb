## Diccionario de Variables

| Columna | Descripción | Tipo de dato |
|----------|-------------|--------------|
| store_nbr | Identificador único de la tienda | `int64` |
| city | Ciudad donde se encuentra ubicada la tienda | `str` |
| state | Estado o provincia donde se ubica la tienda | `str` |
| type | Tipo o categoría de la tienda | `str` |
| cluster | Identificador del clúster al que pertenece la tienda | `int64` |

### Observaciones

- **store_nbr** es el identificador único de cada tienda y será utilizado para relacionar este dataset con el historial de ventas (`train`).
- **city** y **state** son variables categóricas que permiten realizar análisis geográficos y regionales.
- **type** clasifica las tiendas según su categoría, lo que permitirá evaluar posibles diferencias en el comportamiento de las ventas.
- **cluster** representa una agrupación de tiendas con características similares. Aunque está almacenado como `int64`, corresponde a una variable categórica y no debe interpretarse como un valor numérico continuo.

- El dataset no presenta valores nulos, por lo que inicialmente no se requieren procesos de limpieza relacionados con datos faltantes.

## Análisis de las variables numéricas

A continuación se presentan las principales observaciones obtenidas a partir de las estadísticas descriptivas de las variables numéricas del dataset **Stores**.

### Observaciones

- **store_nbr**
  - Existen **54 tiendas**, identificadas con valores consecutivos entre **1** y **54**.
  - Cada identificador corresponde a una tienda única.
  - Aunque está almacenada como un número entero (`int64`), representa una **variable categórica** y no una magnitud numérica.

- **cluster**
  - Los clústeres toman valores entre **1** y **17**, indicando que las tiendas han sido agrupadas en **17 segmentos** según características comunes definidas por el negocio.
  - La mediana es **8.5**, mientras que el promedio es **8.48**, lo que sugiere una distribución relativamente equilibrada de los identificadores de los clústeres.
  - Al igual que `store_nbr`, esta variable está codificada numéricamente, pero representa una **categoría** y no un valor continuo.
  - Las estadísticas descriptivas (media, desviación estándar y cuartiles) únicamente describen la distribución de los identificadores y **no deben interpretarse como una medida cuantitativa del negocio**.

### Conclusiones

- Las variables **`store_nbr`** y **`cluster`** deben tratarse como **variables categóricas** durante el análisis exploratorio y el modelado.

- No se observan valores fuera de los rangos esperados para los identificadores de tiendas ni para los clústeres.

**El dataset presenta una estructura consistente para las variables numéricas, por lo que no se evidencian anomalías que requieran acciones de limpieza en esta etapa.**
